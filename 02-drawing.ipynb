## Drawing on Image

```python
import cv2

def load_image(path):
    image = cv2.imread(path)
    return image

def show_image(title, image_type):
    cv2.imshow(title,image_type)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

> Drawing Line
- pt1 = (100, 100) → start point (x=100, y=100)
- pt2 = (735, 100) → end point (x=735, y=100)
- Thickness:
  - -1: Fill
  - 1+ : px
---

```python
image = load_image("assets\example.JPG")

pt1 = (100, 100)
pt2 = (735, 100)

line = cv2.line(image, pt1, pt2, (255,0,255), 3)

show_image("Github: ShivamDevHere", line)
```

> Drawing Rectangle
---

```python
image = load_image("assets\example.JPG")

pt1 = (10, 10)
pt2 = (700, 700)

rectangle = cv2.rectangle(image, pt1, pt2, (255,0,255), 3)

show_image("Github: ShivamDevHere", rectangle)
```

> Drawing Circle
---

```python
image = load_image("assets\example.JPG")

h, w, c = image.shape
center = (w//2)+40, (h//2)+40

circle = cv2.circle(image, center, 300, (255,0,255), -1)

show_image("Github: ShivamDevHere", circle)
```

> Text over Image
---

```python
image = load_image("assets\example.JPG")

text = cv2.putText(image, "Portfolio: https://portfolio-ruby-one-66.vercel.app/", (50,680), cv2.FONT_HERSHEY_COMPLEX, 1, (0,50,200), 1) 

show_image("Github: ShivamDevHere", text)
```
