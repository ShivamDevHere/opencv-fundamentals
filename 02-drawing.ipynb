## Drawing on Image

```python
import cv2

def load_image(path):
    image = cv2.imread(path)
    return image

def show_image(title, image_type):
    cv2.imshow(title,image_type)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

> Drawing Line
- pt1 = (100, 100) → start point (x=100, y=100)
- pt2 = (735, 100) → end point (x=735, y=100)
---

```python
image = load_image("assets\example.JPG")

pt1 = (100, 100)
pt2 = (735, 100)

line = cv2.line(image, pt1, pt2, (255,0,255), 3)

show_image("Github: ShivamDevHere", line)
```
