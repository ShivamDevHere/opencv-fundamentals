> Resizing and Scaling Image

> Cropping Using Slice

> Image Rotation
- rotate = cv2.getRotationMatrix2D(center, angle, scale)
- rotated = cv2.warpAffine(image, rotate, (w,h))
---

```python

image = cv2.imread("assets\example.JPG")

h, w, c = image.shape
center = w//2, h//2

rotate = cv2.getRotationMatrix2D(center, 300, 1.0)
rotated = cv2.warpAffine(image, rotate, (w,h))

cv2.imshow("Github: ShivamDevHere", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

> Image Flipping
- 0: Verticle
- 1: Horizontal
- -1: Both
---

```python
flipped = cv2.flip(image, -1) 

cv2.imshow("Github: ShivamDevHere", flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
