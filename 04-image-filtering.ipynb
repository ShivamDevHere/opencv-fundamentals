> Gaussian Blur

- `blurred_image = cv2.GaussianBlur(image, (kernel_size_x, kernel_size_y), sigma)`
  - Kernel_size: (1,1) to (21,21) only odd low to high blur
  - sigma: 0 is auto shows blur strongness
---

> Median Blur
    - `blurred = cv2.medianBlur(image, kernel_size)`
