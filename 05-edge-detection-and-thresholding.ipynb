> Canny Edge Detection
---

  - `edges = cv2.canny(img, threshold 1, threshold 2)`
      - In simple: value <= T1 <= T2 <= value
      - Value: edges or details of image will be neglected

```python
import cv2

def canny():
    image = cv2.imread("assets\example.JPG", cv2.IMREAD_GRAYSCALE)
    edges = cv2.Canny(image, 50, 150)

    cv2.imwrite("assets\saves\canny.png", edges)
    cv2.imshow("Canny",edges)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

canny()
```

> Threshold Detection
---

```python
import cv2

def threshold():
    image = cv2.imread("assets\example.JPG", cv2.IMREAD_GRAYSCALE)
    # edges = cv2.Canny(image, 50, 150)

    ret, thres = cv2.threshold(image, 220, 255, cv2.THRESH_BINARY)
    cv2.imwrite("assets\saves\\threshold.png", thres)
    cv2.imshow("Threshold",thres)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

threshold()
```

> Bitwise Operation
---

```python
import numpy as np

img1 = np.zeros((300,300), dtype="uint8")
img2 = np.zeros((300,300), dtype="uint8")

cv2.circle(img1, (150,150), 100, 255, -1)
cv2.rectangle(img2, (100,100), (250,250), 255, -1)

bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or = cv2.bitwise_or(img1, img2)
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow("Circle",img1)
cv2.imshow("Rectangle",img2)
cv2.imshow("AND: Intersect",bitwise_and)
cv2.imshow("OR: Join",bitwise_or)
cv2.imshow("NOT: Inversion",bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows()


```
