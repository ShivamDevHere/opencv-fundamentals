## Video Related Operations

```python
import cv2
```

> Capturing Video
---

```python
def capture_show():
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Coudn't Read Frame")
            break

        cv2.imshow("Webcame", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Exit")
            break

    cap.release()
    cv2.destroyAllWindows()

capture_show()
```
