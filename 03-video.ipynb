## Video Related Operations

```python
import cv2
```

> Capturing Video
---

```python
def capture_show():
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Coudn't Read Frame")
            break

        cv2.imshow("Webcame", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("Exit")
            break

    cap.release()
    cv2.destroyAllWindows()

capture_show()
```

> Saved Video Frames

- `cv2.VideoWriter(filename, codec, fps, frame_size)`
---


```python
camera = cv2.VideoCapture(0)

frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

codec = cv2.VideoWriter_fourcc(*'XVID')
recorded = cv2.VideoWriter("assets\saves\saved_video.avi", codec, 20, (frame_width, frame_height))

while True:
    success, image = camera.read()

    if not success:
        break
    recorded.write(image)
    cv2.imshow("Recording Live", image)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

camera.release()
recorded.release()
cv2.destroyAllWindows()
```
