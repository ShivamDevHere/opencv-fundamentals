## Contours & Shape Detection

- **Contours:** Detects the boundaries of objects in an image using their outlines.
- **Shape Detection:** Identifies geometric shapes such as circles, triangles, rectangles, and squares by analyzing their contours.
    
- Use:
  - Finding Contours
  - Shape Detection
  - Drawing Contours

```python
    cv2.findContours(image, mode, method)
    cv2.drawContours(image, contours, contours_index, color, thickness)
    
```
  - image: Must be only black and white
  - Contours_index:
    - 0: One Shape
    - 1: Second Shape
    - -1: All Shapes


````python
import cv2

def load_image(path):
    image = cv2.imread(path)
    return image

def show_image(title, image_type):
    cv2.imshow(title,image_type)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
````

> Finding Contours
---

````python
def finding_contours():
    image = load_image("assets\\threshold.png")

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

    contours, heirarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    cont = cv2.drawContours(image, contours, -1, (0,255,0), 3)

    show_image("Contoured", cont)

finding_contours()
````

````python
def finding_contours():
    image = load_image("assets\\triangle.png")

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 210, 255, cv2.THRESH_BINARY)

    show_image("Thres", thresh)

    contours, heirarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    cont = cv2.drawContours(image, contours, -1, (0,255,0), 3)

    for contour in contours:

        approx = cv2.approxPolyDP(contour, 0.05 * cv2.arcLength(contour,True),True)

        corners = len(approx)

        if  corners==3:
            shape_name = "Triangle"
        elif corners==4:
            shape_name = "Quadlatertal"
        elif corners==5:
            shape_name = "Pentagon"
        else:
            print("Unknown Shape")

        cv2.drawContours(image, [approx], 0, (0,255,0), 2)
        x = approx.ravel()[0]
        y = approx.ravel()[-1]-10
        cv2.putText(image, shape_name, (x,y), cv2.FONT_HERSHEY_TRIPLEX, 0.6, (255,0, 0), 2)

        show_image(f"Shape: {shape_name}",image)
        
finding_contours()

    
````
